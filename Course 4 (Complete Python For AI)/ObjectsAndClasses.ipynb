### Real World Example- Speed Controll Mechanizer To Maintain The Car Within Speed Limit

```python
class Car:
    mx_speed=140
    def __init__(self,make,model,color,speed=0):
        self.make=make
        self.model=model
        self.color=color
        self.speed=speed
    def accelarate(self):
        if self.speed<=Car.mx_speed:
            acceleration=Car.mx_speed-self.speed
            self.speed=self.speed+acceleration
            print("You Can Accelerate By = ",acceleration)
            print("After Acceleration By",acceleration,"=",self.speed)
        else:
            deccelaration=self.speed-Car.mx_speed
            self.speed=self.speed-deccelaration
            print("Immediately Deccelerate By = ",deccelaration)
            print("After Decceleration By",deccelaration,"=",self.speed)
    def get_speed(self):
        return self.speed

        
n=int(input("What Is Your Car Speed= "))
Car1=Car("Marcedes","Benz","Blue",n)
Car1.accelarate()
print(Car1.get_speed())
print("Speed Limit Is= ",Car1.mx_speed)
```

## Import A Library To Draw The Objects

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
class Circle(object):

    # Constructor
    def __init__(self,radius=3,color="blue"):
        self.radius=radius
        self.color=color

    # Method
    def add_radius(self,r):
        self.radius=self.radius+r
        return (self.radius)

    # Method
    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0,0),radius=self.radius,fc=self.color))
        plt.axis("scaled")
        plt.show()
C2=Circle(3,"blue")
C2.drawCircle()
```

```python
C2.add_radius(2)
C2.drawCircle()
```

```python
C2.radius
```

```python
C2.color="orange"
C2.drawCircle()
```

```python
class Rectangle(object):

    # Constructor
    def __init__(self,width,height,color):
        self.height=height
        self.width=width
        self.color=color

    # Method
    def drawRectangle(self):
        plt.gca().add_patch(plt.Rectangle((0,0),self.width,self.height,fc=self.color))
        plt.axis("scaled")
        plt.show()
R1=Rectangle(25,16,"yellow")
R1.drawRectangle()
```

```python
print(R1.width)
print(R1.height)
R1.color="orange"
R1.drawRectangle()
```

```python
class Vehicles:
    color="white"
    def __init__(self,mx_speed,mileage):
        self.mx_speed=mx_speed
        self.mileage=mileage
    def seating(self,seats):
        self.seats=seats
    def display(self):
        print("Maximum Speed=",self.mx_speed,"KMPH")
        print("Mileage=",self.mileage,"KMPL")
        print("Seats=",self.seats)
        print("Default Color=",self.color)
```

```python
V1=Vehicles(200,20)
V1.seating(5)

V2=Vehicles(180,25)
V2.seating(4)

print("Detailes Of First Vehicle:")
V1.display()

print("Details Of Second Vehicle:")
V2.display()
```
