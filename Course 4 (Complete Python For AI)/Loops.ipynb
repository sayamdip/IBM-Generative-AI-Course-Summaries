## For Loops

```python
list1=[1,2,3,4,5]
for i in list1:
    print(i)
```

```python
for i in range(0,10,+1):
    print(i)
```

```python
for j in range(10,0,-1):
    print(j)
```

```python
## Multiplication Table
for i in range(1,11,+1):
    for j in range(1,11,+1):
        print(i,"X",j,"= ",i*j)
```

## enumerate

```python
squares=["Red","Yellow","Green"]
for i,square in enumerate(squares):
    print(square)
    print(i)
```

## While Loops

```python
## Copying Non-Orange Elements To The Newsquares List
squares=["Orange","Orange","Purple","Orange","Blue"]
Newsquares=[]
i=0
while(i<len(squares)):
    if(squares[i]!="Orange"):
        Newsquares.append(squares[i])
    i=i+1
print(Newsquares)
```

```python
i=1
j=1
while(i<11):
    while(j<11):
        print(i,"X",j,"= ",i*j)
        j=j+1
    i=i+1
```

```python
dates=[1982,1980,1973]
N=len(dates)
for i in range(N):
    print(dates[i])
for year in dates:
    print(year)
```

```python
squares=["Red","Yellow","Green","Purple","Blue"]
for i in range(0,5):
    print("Before Square ",i," Is ",squares[i])
    squares[i]="White"
    print("After Square ",i," Is ",squares[i])
```

```python
squares=["Red","Yellow","Green","Purple","Blue"]
for i,square in enumerate(squares):
    print(i,square)
```

```python
dates=[1982,1980,1973,2000]
i=0
year=dates[0]
while(year!=1973):
    print(year)
    i=i+1
    year=dates[i]
print("It Took ",i," Repititions To Get Out Of Loop")
```

## Break Statement

```python
for num in range(1,10):
    if num==5:
        print("Breaking The Loop At:",num)
        break
    print(num)
```

## Continue Statement

```python
for num in range(1,10):
    if num==5:
        print("Breaking The Loop At:",num)
        continue
    print(num)
```
