## Strings

#### Strings Are Immutable- Means We Cannot Change The Value Of The String

```python
name="Michael Jackson"
print(name[0:4])
print(name[8:14])
```

#### Find Lenght Of A String Using len

```python
print(len(name)) # Prints The Length Of A String
```

1) String Slicing

```python
print(name[::2]) # Prints The Character And Jumping Two Letters
print(name[0:5:2])
```

2) String Concatenation 

```python
student=name+" Is The Best"
print(student)
```

3) String Replication

```python
print(4*name)
```

4) New Line (\n)

```python
print("Michael Jackson \n Is The Best")
```

## String Methods

#### 1) upper()

```python
a="Thriller is the sixth studio album"
print(a.upper())
```

#### 2) replace()

```python
b="Michael Jackson Is The Best"
print(b.replace("Michael","Janet"))
```

#### 3) find()

```python
print(b.find("Jack")) # It Points To The Index Of The First Character Of A String Sequence To Find
```

#### 4) split()

```python
print(b.split())
```

#### 5) lower()

```python
print(a.lower())
```

#### 6) capitalize()

```python
print(a.capitalize())
```

#### 7) title()

```python
print(a.title())
```

## Format Strings In Python

```python
k="John"
age=40

# Method 1
print(f"My Name Is {k} And I Am {age} Years Old.")

# Method 2
print("My Name Is {} And I Am {} Years Old.".format(k,age))

# Method 3
print("My Name Is %s And I Am %d Years Old."%(k,age))
```

## RegEx (Regular Expression)

### 1) search

```python
import re
s1="The BodyGuard Is The Best Album"
pattern=r"Body"
result=re.search(pattern,s1)
if(result):
    print("Match Found")
else:
    print("Match Not Found")
```

### 2) group

```python
pattern=r"\d\d\d\d\d\d\d\d\d\d"
text="My Phone Number Is 1234567890"
match=re.search(pattern,text)
if(match):
    print("Phone Number Found: ",match.group())
else:
    print("Phone Number Not Found")
```

### 3) findall

```python
pattern=r"\W" # Matches Any Non-Word Character, Example- Special Characters, @, Commas, Space, Etc.
text="Hello, World!@"
matches=re.findall(pattern,text)
print("Matches:",matches)

result=re.findall("Wo",text)
print(result)
```

### 4) split

```python
split=re.split("\s",text)
print(split)
```

### 5) sub

```python
s4="The BodyGuard Is The Best Album Of 'Whitney Houston'."
pattern=r"Whitney Houston"
replacement="Legend"
new_string=re.sub(pattern,replacement,s4,flags=re.IGNORECASE) # re.IGNORECASE makes the search case-insensitive, so it matches "Whitney Houston" in any letter case
print(new_string)
```
