## Sets- Set Only Considers Unique Elements Not Repititive Elements

```python
set1={"Pop","Rock","Soul","Hard Rock","Rock","R&B","Rock","Disco"}
print(set1)
```

```python
# Converting List To A Set
album_list=["Michael Jackson","Thriller","Thriller",1982]
album_set=set(album_list)
print(album_set)
```

#### 1) add()

```python
album_set.add("AC/DC")
print(album_set)
```

#### 2) remove()

```python
album_set.remove("AC/DC")
print(album_set)
```

#### 3) in

```python
print("Who" in album_set)
```

#### 4) And (&)

```python
album_set2={"Thriller","AC/DC","Whitney Houston",1982}
album_set3=album_set & album_set2
print(album_set & album_set2) # It Gives The Values Which Are Present In Both The Sets
```

#### 5) union

```python
print(album_set.union(album_set2))
```

#### 6) issubset

```python
print(album_set3.issubset(album_set))
```

#### 7) difference

```python
print(album_set.difference(album_set2))
```

#### 8) intersection

```python
print(album_set.intersection(album_set2))
```

#### 9) issuperset

```python
album_set.issuperset(album_set2)
```
