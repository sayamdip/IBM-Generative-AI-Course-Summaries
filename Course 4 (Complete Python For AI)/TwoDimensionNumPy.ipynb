### Two Dimensional NumPy

```python
import numpy as np
a=[[11,12,13],[21,22,23],[31,32,33]]
A=np.array(a)
print(A)
```

```python
print(A.ndim)
```

```python
print(A.shape)
```

```python
print(A.size)
```

```python
print(A[0,0:2])
```

```python
a=[[1,0],[0,1]]
b=[[2,1],[1,2]]
p=np.array(a)
o=np.array(b)
print(p+o)
```

```python
print(2*o)
```

```python
print(p*o)
```

```python
print(np.multiply(p,o))
```

```python
print(np.dot(p,o))
```

### Exercise 

```python
a=[[11,12,13],[21,22,23],[31,32,33]]
matrix1=np.array(a)
print(matrix1)
```

```python
# Access the element on the second row and third column
print(matrix1[1][2])
```

```python
# Access the element on the first row and first and second columns
print(matrix1[0][0:2])
```

```python
# Access the element on the first and second rows and third column
print(matrix1[0:2,2])
```
