### Reading XML File Format

```python
import xml.etree.ElementTree as etree
tree=etree.parse("./Cricket.xml")
root=tree.getroot()
teams_data=[]
for team in root.findall(".//team"):
    teams_data.append({
        "Team ID":team.find("teamId").text,
        "Team Name":team.find("teamName").text,
        "Country":team.find("country").text,
        "Coach":team.find("coach").text,
        "Win":team.find(".//winPercentage").text
    })
df=pd.DataFrame(teams_data)
df
```
