```python
customers2={"OrderID":[1,2,3,4,5],"CustomerName":["Sam","Angelina","Mathew","Michael","Lucas"]}
df2=pd.DataFrame(customers2)
df2.head()
```

### Merge Two Data Frames

```python
merge_df=pd.merge(df2,df1,on=["OrderID","OrderID"])
merge_df.head()
```

### tail() Method
Displays the last n rows of the DataFrame.

```python
merge_df.tail(2)
```

### replace()
Replaces specific values in a column with new values.

```python
merge_df=merge_df["CustomerName"].replace("Lucas","Leonardo",inplace=True)
```
