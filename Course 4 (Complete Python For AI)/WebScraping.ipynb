```python
import pandas as pd
from urllib.request import Request, urlopen

URL="https://en.wikipedia.org/wiki/List_of_largest_banks"
req = Request(URL, headers={'User-Agent': 'Mozilla/5.0'})  ## This Line Is Very Important, Otherwise It Will Give Errors
tables = pd.read_html(urlopen(req))
df = tables[0]
df=pd.DataFrame(df)
df
```
