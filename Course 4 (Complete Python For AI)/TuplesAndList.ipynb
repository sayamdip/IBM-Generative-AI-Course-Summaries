## Tuples 
#### Tuples Are Immutable- We Cannot Change The Value Of A Tuple

```python
tuple1=("disco",10,1.2)
print(tuple1[0])
```

```python
# Concatenate Tuples
tuple2=tuple1+("hard rock",10)
print(tuple2)
```

```python
# Slicing In Tuple
print(tuple2[3:5])
```

```python
# Length Of A Tuple
print(len(tuple2))
```

```python
# Nesting In Tuple
NT=(1,2,("pop","rock"),(3,4),("disco",(1,2)))
print(NT)
```

```python
# Sorting In Tuple
C_tuple=(-5,1,-3)
print(sorted(C_tuple))
```

## Lists
#### Lists Are Mutable- We Can Change The Value Of A List

```python
list1=["Michael Jackson",10.11982,[1,2],('A',1)]
print(list1[2])
```

```python
# Slicing In List
print(list1[2:4])
```

#### append() & extend()

```python
# Concatenate
list2=list1+["pop",10]
print(list2)
print(list2.extend(["Rick",80])) # extend Method- It Adds A List To Another List
print(list2.append("O")) # append Method- It Adds Only One Item To A List
print(list2)
```

```python
# List Mutability Example

list1=["Michael Jackson",10.11982,[1,2],('A',1)]
print("Before Change= ",list1)

list1[0]="Whitney Houston"
print("After Change= ",list1)
```

```python
# Slicing
print("Whitney Houston".split(" "))
```

```python
# Delete The Element Based On The Index
del(list1[1])
print(list1)
```

#### copy() & count()

```python
# copy()
my_list=[1,2,3,4]
new_list=my_list.copy()
print(new_list)

# count()
print(new_list.count(2))
```

#### insert() & sort()

```python
# insert()
new_list.insert(3,9)
print(new_list)

# sort()
print(new_list.sort())
print(new_list)
print(new_list.sort(reverse=True))
print(new_list)
```

#### sum(), min() & max()

```python
print(sum(my_list))
print(min(my_list))
print(max(my_list))
```

#### reverse() & remove()

```python
print(my_list)

# reverse()
my_list.reverse()
print(my_list)

# remove()
my_list.remove(1)
print(my_list)
```
