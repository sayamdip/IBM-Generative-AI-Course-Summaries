## Data Types In Python

1) Integers

```python
a=40
print(a)
```

2) Floats

```python
b=40.9
print(b)
```

3) Boolean

```python
o=True
print(o)
```

## Use Of type Keyword

```python
print(type(4))
print(type(4.5))
print(type(True))
```

## Converting One Data Type To Another

# Integers To Floats

```python
print(float(4))
print(type(float(4)))
```

# Strings To Integers Or Floats

```python
print(int('4'))
print(type(int('4')))

print(float('4.5'))
print(type(float((4.5))))
```

# Numbers To Strings

```python
print(str(4))
print(type(str(4)))
print(str("123-456-7890"))
```

## Boolean To Strings

```python
print(int(True))
print(int(False))
```
