## Expressions And Variables

### Expressions

```python
# Addition operation Expression
print(43+60+16+41)
```

```python
# Subtraction Operation Expression
print(50-60)
```

```python
# Multiplication Operation Expression
print(5*5)
```

```python
# Division Operation Expression
print(25/6)
```

```python
# Floor Division Expression
print(25//6)
```

### Variables

```python
a=43+60+16+41
print(a)

b=50-60
print(b)

o=5*5
print(o)

e=25/6
print(e)
```
