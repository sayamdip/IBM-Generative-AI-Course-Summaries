## Dictionaries
1) Dictionaries Are Denoted With Curly Brackets {}.
2) The Keys Have To Be Immutable And Unique.
3) The Values Can Be Immutable, Mutable And Duplicates.

```python
dict1={"Thriller":1982,"Back In Black":1980,"The Dark Side Of The Moon":1973,"The Bodyguard":1992}
print(dict1)
```

```python
# Accessing The Values Of Keys
print(dict1["Thriller"])
print(dict1["Back In Black"])
print(dict1["The Dark Side Of The Moon"])
print(dict1["The Bodyguard"])
```

```python
# Adding Value To A Dictionary
dict1["Graduation"]=2007
print(dict1)
```

```python
# Deleting Values In  Dictionary
del(dict1["Thriller"])
print(dict1)
```

```python
# Check Whether A Key Is In A Dictionary Or Not
print("Thriller" in dict1)
```

```python
# Obtaning Values In A Dictionary
print(dict1.values())
```

```python
# Creating A Dictionary
dict2={"key1":1,"key2":2,"key3":[3,3,3],"key4":(4,4,4),"key5":5,(0,1):6}
print(dict2)
```

```python
# Check The Keys In A Dictionary
print(dict2.keys())
```
