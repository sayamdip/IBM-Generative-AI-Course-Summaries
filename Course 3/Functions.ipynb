#### List Operations With Function

##### 1) add_element Function

```python
def add_element(list1,element):
    list1.append(element)
list1=[]
add_element(list1,10)
add_element(list1,9)
add_element(list1,6)
add_element(list1,7)
add_element(list1,8)
print(list1)
```

##### 2) remove_element Function

```python
def remove_element(list1,element):
    if element in list1:
        list1.remove(element)
        print(element,"Is Removed")
    else:
        print(element," Not Found In The List")
remove_element(list1,8)
remove_element(list1,5)
print(list1)
```

```python
def count1(str1, word):
    count = 0
    words = str1.split()  # Split string into list of words
    for w in words:
        if w == word:
            count = count + 1
    print(count)

str1 = "Mary had a little lamb Little lamb, little lamb Mary had a little lamb.Its fleece was white as snow And everywhere that Mary went Mary went Everywhere that Mary went The lamb was sure to go"
word = "little"
count1(str1, word)
```
