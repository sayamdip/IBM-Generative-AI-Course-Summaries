## Text Analysis

```python
givenstring="Lorem ipsum dolor! diam amet, consetetur Lorem magna. sed diam nonumy eirmod tempor. diam et labore? et diam magna. et diam amet."
```

```python
class TextAnalyzer(object):
    def __init__(self,text):
        frmt_text=text.replace('.','').replace(',','').replace('!','').replace('?','')
        frmt_text=frmt_text.lower()
        self.frmt_text=frmt_text
        print("In Lowercase=",self.frmt_text)
    def freqAllandOff(self,wordcheck):
        wordList=self.frmt_text.split(' ')
        freq={}
        for word in set(wordList):
            freq[word]=wordList.count(word)
        print(freq)
        print(wordcheck,'Appears=',freq[wordcheck],'Times')
```

```python
A1=TextAnalyzer(givenstring)
A1.freqAllandOff('lorem')
```
