### Chi-Square Test

```python
# Example- Weak Correlation
import pandas as pd
from scipy.stats import chi2_contingency

# Create The Contingency Table
data=[[20, 30], # Male: [Like, Dislike]
      [25, 25]] # Female: [Like, Dislike]

# Create A Dataframe
df = pd.DataFrame(data, columns=["Like", "Dislike"], index=["Male", "Female"])

# Perform The Chi-Square Test
chi2, p, dof, expected = chi2_contingency(df, correction=False)

# Display Results
print("Chi-square Statistics:", chi2)
print("Degrees Of Freedom:", dof)
print("P-Value:",p)
print("Expected Frequencies:\n",expected)
```
