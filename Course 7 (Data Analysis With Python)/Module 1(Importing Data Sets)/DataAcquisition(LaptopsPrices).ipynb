```python
import pandas as pd
import numpy as np
```

```python
import requests

url= "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/laptop_pricing_dataset_base.csv"
response = requests.get(url)

with open("laptop.csv", "wb") as f:
    f.write(response.content)
```

```python
file_path = "laptop.csv"
df=pd.read_csv(file_path)
df.head(10)
```

```python
## Create A Headers List
headers=["Manufacturer", "Category", "Screen", "GPU", "OS", "CPU_core", "Screen_Size_inch", "CPU_frequency", "RAM_GB", "Storage_GB_SSD", "Weight_kg", "Price"]
```

```python
df.columns = headers
df
```

```python
## Replace '?' with 'NaN'

df1 = df.replace("?", np.NaN)
df 
```

```python
## Print the statistical description of the dataset, including that of 'object' data types

print(df1.describe())
```
